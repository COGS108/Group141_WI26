## Abstract

This project investigates whether neighborhood socioeconomic resources are associated with civic participation. Specifically, we ask whether **median household income** in a California census tract is associated with **voter participation** in recent election cycles.

We analyze a tract-level dataset that merges (1) cross-walked turnout estimates and registration counts with (2) American Community Survey (ACS) 5-year median household income. The final analytic file contains **3,958 tract–year observations** across **2,452 unique census tracts** for **2022** and **2024**. We examine the income–participation relationship using exploratory visualization, Pearson/Spearman correlations, and linear regression with tract-clustered standard errors. We study **turnout among registered voters** and, by decomposition, both **registration rates** and **turnout using the adult-population denominator**.

Across both years, higher-income tracts show substantially higher participation. In our primary regression, a **$10,000** increase in tract median household income is associated with about **2.25 percentage points** higher turnout among registered voters (95% CI [2.16, 2.34]) and about **3.31 percentage points** higher turnout among the adult denominator (95% CI [3.19, 3.44]). These results are consistent with the idea that socioeconomic resources are linked to both **registration** and **voting conditional on registration**, though we emphasize the findings are **associational**, not causal.

## Research Question

To what extent is neighborhood-level median household income (at the census tract level) associated with voter participation across California in recent election cycles (2022 and 2024)?

## Background and Prior Work

<p>
<a id="ref1"></a>
Voter turnout is a core measure of democratic participation, but participation in elections is highly unequal across socioeconomic groups in the United States: richer people are more likely to vote than poorer people.<sup><a href="#fn1">1</a></sup>
Recent U.S. elections have also seen unusually high turnout by historical standards. For example, Pew Research Center reports that 2020 turnout reached levels not seen in decades, estimating that 62.8% of the voting-age population voted, while noting that U.S. turnout still trails many peer democracies under common cross-national turnout measures.<sup><a href="#fn2">2</a></sup>
</p>

<p>
<a id="ref2"></a>
Income-based inequalities in participation are important because they can affect political representation. When lower-income citizens vote at lower rates, their interests and preferences are less likely to be reflected in public policy, even if income itself is not necessarily the direct cause of voting behavior.<sup><a href="#fn1">1</a></sup>
At the same time, concerns about unequal participation are heightened in a period of rising economic inequality in the United States; Inequality.org summarizes that the top 1% share of national income has increased substantially since 1970 (citing the World Inequality Database).<sup><a href="#fn3">3</a></sup>
</p>

<p>
<a id="ref3"></a>
Prior research has examined how income inequality relates to turnout and to “income bias” in turnout (the degree to which higher-income people vote more than lower-income people), but has reached mixed conclusions.
Matsubayashi and Sakaiya (2021), using large-scale cross-national survey data, argue that higher income inequality can be associated with lower income bias in turnout by demobilizing higher-income citizens while mobilizing lower-income citizens, and they discuss potential mechanisms such as vote buying.<sup><a href="#fn4">4</a></sup>
At the same time, more recent precinct-level work emphasizes that a major obstacle in this literature is measurement: survey-based turnout can be affected by response and social desirability bias, which can produce widely varying estimates of income-based turnout gaps across studies.<sup><a href="#fn5">5</a></sup>
</p>

<p>
<a id="ref4"></a>
Because voter participation varies across communities, our project examines whether California census tracts with higher median household incomes tend to exhibit systematically different voter turnout rates. A tract-level design allows us to study neighborhood-level socioeconomic context more directly than county-level aggregation, while also acknowledging that these remain ecological, not individual-level, associations.1,2,5<sup><a href="#fn1">1</a></sup><sup><a href="#fn2">2</a></sup>

We hypothesize that California census tracts with higher median household income will have higher voter participation.

## Data

### Data overview

**Unit of analysis.** California **census tract** (tract–year observations).

**Dataset name and sources.** *California Neighborhood Voting Rates (Cross-walked to 2023 ACS 5-Year Estimates)*, constructed by integrating:
- California voter participation data (tract-level, cross-walked from election geography): https://statewidedatabase.org/
- U.S. Census Bureau **ACS 5-year** estimates (income): https://www.census.gov/data/developers/data-sets/acs-5year.html

**Why tract-level?** We initially considered a county-level comparison, but shifted to a more granular, neighborhood-level unit (census tracts) to better capture local socioeconomic context and reduce overly coarse aggregation. Census tracts still represent aggregated data rather than individual-level behavior, but they provide a more localized measure of neighborhood socioeconomic conditions than counties.

**Coverage and size.** The merged dataset contains **3,958** tract–year observations across **2,452** unique census tracts for **2022** and **2024**.

**How the data were combined.** The final analytic file combines tract-level voter participation estimates from the California Statewide Database with median household income from the **2019–2023 ACS 5-year estimates**. Median household income was retrieved through the official U.S. Census Bureau ACS API and merged to the turnout data using standardized tract GEOIDs.

For the final report, the analytic dataset is stored as `data/02-processed/ca_tract_voter_income_final.csv`.

**Key variables.**
- `median_household_income`: ACS 2019–2023 5-year estimate of median annual household income (USD) for the tract.
- `registered_count`: number of registered voters in the tract.
- `voted_count`: number of ballots cast (for the election record used in the turnout file).
- `registration_pct`: `registered_count / denom_adult_pop × 100`.
- `turnout_registered_pct`: `voted_count / registered_count × 100` (provided as `voter_turnout_pct` in the file).
- `turnout_adult_pct` (constructed): `voted_count / denom_adult_pop × 100`
  (equivalently, `registration_pct × turnout_registered_pct / 100`).

**Measurement notes.** The ACS income measure is a **5-year period estimate**, not a single-year snapshot. It should therefore be interpreted as a tract-level socioeconomic proxy for the study period rather than as an exact income measure for one election year. Likewise, tract-level turnout estimates rely on geographic cross-walking because voting precinct boundaries do not perfectly align with census tracts, so some values should be interpreted as best available estimates rather than exact administrative counts.

### Data wrangling

The source turnout file was already mostly tidy, so only limited cleaning was required for those fields. The main wrangling work involved integrating tract-level turnout data with ACS income data, standardizing geographic identifiers, constructing analysis variables, and applying validity checks before analysis.

In the final analysis we:
1. verify numeric ranges and missingness,
2. construct `turnout_registered_pct`, `turnout_adult_pct`, and `income_10k`,
3. exclude tracts with missing or invalid income values (including suppressed Census placeholder codes),
4. constrain percentage variables to the 0–100 range, and
5. remove a small number of impossible adult-denominator turnout values outside the 0–100 range only for analyses using that outcome.

After these checks, the analytic dataset retains **3,958** tract–year observations for the main registered-turnout analyses and **3,949** observations for analyses using the adult-population denominator.

```python
from IPython.display import display
import os
import numpy as np
import pandas as pd
DATA_PATH = "data/02-processed/ca_tract_voter_income_final.csv"
df = pd.read_csv(DATA_PATH)

df.head()

# Standardize / construct variables used in the report
df["turnout_registered_pct"] = df["voter_turnout_pct"]
df["turnout_adult_pct"] = df["voted_count"] / df["denom_adult_pop"] * 100
df["income_10k"] = df["median_household_income"] / 10000

# Validity checks
df = df[df["median_household_income"].notna()].copy()
df = df[df["median_household_income"] > 0].copy()
df = df[df["registration_pct"].between(0, 100)].copy()
df = df[df["turnout_registered_pct"].between(0, 100)].copy()

# Keep a version for adult-denominator turnout analyses (remove rare impossible values)
df_adult = df[df["turnout_adult_pct"].between(0, 100)].copy()

print("Loaded:", DATA_PATH)
print("Rows (all):", len(df), "| Rows (adult-turnout-valid):", len(df_adult))
print("Unique tracts:", df["geoid"].nunique(), "| Years:", sorted(df["year"].unique()))
```

These descriptive patterns suggest that the income–participation relationship is both visually strong and substantively meaningful. We next quantify this relationship more formally using correlations and tract-clustered regression models.

### Income and turnout among registered voters

We first quantify the income–turnout relationship using correlation and linear regression.

- **Correlation (all years pooled):** Pearson *r* ≈ **0.591**; Spearman ρ ≈ **0.622** (both p < 0.001).  
  Because overall turnout differs by year, we also compute within-year correlations, which are similarly strong.

- **Regression (clustered by tract):** We regress `turnout_registered_pct` on median household income, scaled in **\$10,000 units**, and a year indicator. The estimated slope is about **2.25 percentage points** per **\$10,000** (95% CI **[2.16, 2.34]**), with **R² ≈ 0.711**. This indicates a strong positive descriptive association between tract income and turnout among registered voters, though the result should not be interpreted as causal.

**Interpretation.** A tract with $100,000 higher median income is predicted to have roughly **22.5 percentage points** higher turnout among registered voters, holding the year constant. This is a large descriptive difference, but it should not be interpreted as causal because income is correlated with many omitted structural factors.

```python
from scipy.stats import pearsonr, spearmanr
import statsmodels.formula.api as smf

def corr_report(data, x, y):
    pr = pearsonr(data[x], data[y])
    sr = spearmanr(data[x], data[y])
    return {
        "n": len(data),
        "pearson_r": pr.statistic, "pearson_p": pr.pvalue,
        "spearman_rho": sr.correlation, "spearman_p": sr.pvalue,
    }

# Correlations (overall + by year)
corr_overall = corr_report(df, "median_household_income", "turnout_registered_pct")
corr_by_year = {
    int(yr): corr_report(df[df["year"] == yr], "median_household_income", "turnout_registered_pct")
    for yr in sorted(df["year"].unique())
}

print("Correlation (overall):", corr_overall)
print("Correlation (by year):", corr_by_year)

# OLS with tract-clustered SE
m1 = smf.ols("turnout_registered_pct ~ income_10k + C(year)", data=df).fit(
    cov_type="cluster", cov_kwds={"groups": df["geoid"]}
)

display(m1.summary2().tables[1])
print("R^2:", round(m1.rsquared, 3))
```

### Decomposing “turnout among adults”: registration and voting conditional on registration

Turnout among the adult denominator can be expressed as:

$$
\begin{aligned}
T_{\text{adult}}
&= \frac{V}{A}\times 100 \\
&= \left(\frac{R}{A}\times 100\right)\left(\frac{V}{R}\times 100\right)\Big/ 100 .
\end{aligned}
$$

where \(A=\) denom adult population, \(R=\) registered count, \(V=\) voted count.

So differences in participation can come from both (i) higher registration and (ii) higher voting among those registered.

Empirically, income is strongly associated with both components:
- Income vs registration: Pearson *r* ≈ **0.612** (p < 0.001).
- Income vs turnout among registered: Pearson *r* ≈ **0.591** (p < 0.001).

In tract-clustered regressions controlling for year:
- Registration increases by **2.62 percentage points** per $10,000 (95% CI [2.48, 2.77]).

- Turnout among the adult denominator increases by **3.31 percentage points** per $10,000 (95% CI [3.19, 3.44]), R² ≈ **0.641**.

**Year differences.** Overall turnout is higher in 2024 than 2022. When allowing the income slope to differ by year, the income–turnout slope for turnout among registered voters is slightly smaller in 2024 than 2022 (interaction p ≈ 1.33e-22), but remains strongly positive in both years.

```python
# Decomposition outcomes: registration and adult-denominator turnout

m_reg = smf.ols("registration_pct ~ income_10k + C(year)", data=df).fit(
    cov_type="cluster", cov_kwds={"groups": df["geoid"]}
)

m_adult = smf.ols("turnout_adult_pct ~ income_10k + C(year)", data=df_adult).fit(
    cov_type="cluster", cov_kwds={"groups": df_adult["geoid"]}
)

print("Registration model:")
display(m_reg.summary2().tables[1])
print("R^2:", round(m_reg.rsquared, 3))

print("\nAdult-denominator turnout model (filtered to 0–100):")
display(m_adult.summary2().tables[1])
print("R^2:", round(m_adult.rsquared, 3))

# Year-varying slope (interaction)
m_int = smf.ols("turnout_registered_pct ~ income_10k * C(year)", data=df).fit(
    cov_type="cluster", cov_kwds={"groups": df["geoid"]}
)
print("\nInteraction model (income slope differs by year):")
display(m_int.summary2().tables[1])
```

### Robustness check (population-weighted regression)

Because tracts vary in size, we re-estimate the turnout-among-registered model using weighted least squares, weighting by `denom_registered`. The estimated income slope remains very similar to the unweighted model (about 2.17 percentage points per $10,000 in the weighted model versus about 2.25 in the main model), suggesting that the main result is not driven only by small tracts.

```python
# Population-weighted robustness check (weight by registered voters)
m_wls = smf.wls("turnout_registered_pct ~ income_10k + C(year)", data=df, weights=df["denom_registered"]).fit(
    cov_type="cluster", cov_kwds={"groups": df["geoid"]}
)
display(m_wls.summary2().tables[1])
print("R^2:", round(m_wls.rsquared, 3))
```

## Ethics

This project uses only publicly available, aggregated data at the **census-tract** level (California Statewide Database turnout estimates cross-walked from voting precincts to census tracts and merged with ACS 5-year socioeconomic estimates). Because we do not analyze individual-level records, we do not collect, store, or expose personally identifiable information (PII). Nevertheless, tract-level data can still create ethical risks when interpreted or communicated carelessly, so we acknowledge these risks and interpret the results carefully.

**Data collection and measurement.** Our main inputs—ACS 5-year median household income and tract-level turnout rates—are widely used public measures, but both involve uncertainty. ACS estimates are based on survey sampling and self-reporting, and tract-level turnout values rely on geographic cross-walking because voting precinct boundaries do not perfectly align with census tracts. We address these issues by clearly describing the data sources and preprocessing steps, filtering out Census “no data/suppressed” codes (e.g., negative placeholders) rather than treating them as real values, and treating our results as **associational** rather than causal.

**Bias, missingness, and representativeness.** Data availability is not random: tracts with suppressed or missing ACS values are excluded by necessity, and cross-walked turnout may introduce more error in dense or rapidly changing areas. This means our final analytic dataset may overrepresent tracts with **complete and valid ACS and turnout records**. We mitigate this by clearly reporting filtering rules (e.g., `income > 0`, turnout constrained to 0–100) and by emphasizing that the analysis describes the tracts that remain after validity checks, not a perfect census of “all neighborhoods.”

**Fairness and interpretation.** Because income is correlated with many structural factors (education, age composition, housing stability, urbanization, etc.), a bivariate relationship between income and turnout can reflect confounding rather than a direct “income effect.” To avoid misleading narratives or stigmatizing low-income communities, we frame conclusions carefully: we do not claim that lower-income residents are inherently less civic-minded, and we explicitly note alternative explanations and omitted-variable risk. Because the analysis is conducted at the census-tract level, the observed patterns should not be interpreted as individual-level behavior, which would risk committing the **ecological fallacy**.

**Data storage, security, and reproducibility.** Although the data are public and aggregated, we follow basic data integrity practices by maintaining a transparent and reproducible data-processing pipeline. Processed data are stored locally within the project structure, and the finalized dataset is loaded from `data/02-processed` for the final analysis so that the results can be re-run without repeating slow API calls and to reduce the chance of accidental inconsistencies.

**Deployment and downstream use.** This work is conducted for an academic research project and is not deployed as a decision-making system. However, tract-level findings can still be misused (for example, to label communities as “good” or “bad” voters). To reduce this risk, we present results with appropriate uncertainty and limitations, avoid individual-level inference, and discourage causal or moralized interpretations of descriptive patterns.

## Discussion and Conclusion

Our results show a consistent, substantively large positive association between neighborhood median household income and voter participation across California census tracts in both 2022 and 2024. In both descriptive plots and regression models, higher-income tracts have higher turnout among registered voters and higher registration rates.

A key takeaway is that the income–participation relationship appears to reflect differences in **both** components of turnout using the adult-population denominator: (1) registration and (2) voting conditional on registration. This distinction may matter for interpretation, because registration-focused strategies and strategies aimed at mobilizing already-registered voters could relate to participation through different channels.

At the same time, the analysis has important limitations. First, the study is observational, and tract income is correlated with many omitted variables (education, age structure, residential stability, urban–rural context, local campaign contact, and more). Second, the turnout denominator is subject to cross-walk and measurement noise, which we see in a small number of implausible adult-turnout values. Third, the dataset covers California tracts, so patterns may differ in other states with different election administration and demographics.

Despite these limitations, the findings align with prior work on socioeconomic inequality in civic participation: higher-resource communities tend to participate at higher rates. Future work could add tract-level controls (education, housing stability, racial/ethnic composition, urbanization), incorporate additional election years, and explore causal designs (e.g., natural experiments around registration policies) to better distinguish mechanisms.
